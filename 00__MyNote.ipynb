# VS Code 가상환경 생성(WIndows)


1. [Python Download Link](https://www.python.org/downloads/windows/) 
-> Click "Windows installer (64-bit)"

2. Visual Studio Code에서 `Ctrl` + `Shift` + `~` 눌러 터미널 열기

3. Copy and paste commands below on Terminal
```PowerShell
python --version
python -m venv --system-site-packages .venv
.\.venv\Scripts\Activate.ps1
python -m pip install --upgrade pip
pip install jupyter ipykernel
python -m ipykernel install --user --name=vision_llm --display-name "Vision-LLM (Python3.10)"
```

4. Close VS Code and Restart
