# Data Types

````python
# Data Types: int, float, str, bool, list, tuple, dict, set
intint = 1; floatfloat = 1.0; strstr = "12321"; boolbool = True
listlist = [1, 2, 3]; tupletuple = (1, 2, 3); dictdict = {"key": "value"}; setset = {1, 2, 3}

# Convert between data types
int_to_float = float(intint); float_to_int = int(floatfloat)
str_to_int = int(strstr); str_to_float = float(strstr)
int_to_str = str(intint); float_to_str = str(floatfloat)
bool_to_int = int(boolbool); int_to_bool = bool(intint)

# Check data types
print(f"int: {intint}, float: {floatfloat}, str: {strstr}, bool: {boolbool}")
print(f"list: {listlist}, tuple: {tupletuple}, dict: {dictdict}, set: {setset}")
print(f"int to float: {int_to_float}, float to int: {float_to_int}")
print(f"str to int: {str_to_int}, str to float: {str_to_float}")
print(f"int to str: {int_to_str}, float to str: {float_to_str}")
print(f"bool to int: {bool_to_int}, int to bool: {int_to_bool}")

# Length of data types
print(f"Length of int: {len(str(intint))}, float: {len(str(floatfloat))}, str: {len(strstr)}, bool: {len(str(boolbool))}")
print(f"Length of list: {len(listlist)}, tuple: {len(tupletuple)}, dict: {len(dictdict)}, set: {len(setset)}")
````

# Slicing

````python
# 1. Supported: Sequence types
text = "Python"
num_list = [10, 20, 30, 40, 50]
num_tuple = (10, 20, 30, 40, 50)
range_obj = range(10)

print("String slicing:", text[1:4])          # Output: 'yth'
print("List slicing:", num_list[1:3])        # Output: [20, 30]
print("Tuple slicing:", num_tuple[::2])      # Output: (10, 30, 50)
print("Range slicing:", list(range_obj[2:7]))  # Output: [2, 3, 4, 5, 6]

# 2. Unsupported: Scalar types (int, float, bool)
number = 12345
try:
    _ = number[0:2]
except TypeError as error:
    print("\nScalar error (int):", error)
    # Output: 'int' object is not subscriptable

# 3. Unsupported: Unordered collection (set)
num_set = {10, 20, 30, 40, 50}
try:
    _ = num_set[0:2]
except TypeError as error:
    print("Set error:", error)
    # Output: 'set' object is not subscriptable

# 4. Unsupported: Mapping type (dict)
num_dict = {"a": 1, "b": 2, "c": 3}
try:
    _ = num_dict[0:2]
except TypeError as error:
    print("Dict error:", error)
    # Output: unhashable type: 'slice'
````

# Lists VS Tuples

````python
import sys

# 1. Mutability vs Immutability
sample_list = [1, 2, 3]
sample_tuple = (1, 2, 3)

sample_list[0] = 99  # Valid: List is mutable
print("Modified list:", sample_list)

try:
    sample_tuple[0] = 99  # Raises TypeError: Tuple is immutable
except TypeError as error:
    print("Tuple modification error:", error)

# 2. Memory Consumption Comparison
print("List memory size (bytes):", sys.getsizeof(sample_list))
print("Tuple memory size (bytes):", sys.getsizeof(sample_tuple))

# 3. Hashability (Usage as Dictionary Key)
coordinates_dict = {
    (37.5665, 126.9780): "Seoul",  # Valid: Tuple can be a dictionary key
}
print("Dictionary value with tuple key:", coordinates_dict[(37.5665, 126.9780)])
````

# Enumerate, Zip

````python
students = ["짱구", "철수", "맹구"]
scores = [85, 92, 78]

print(enumerate(students))
print(tuple(enumerate(students)))
print(list(enumerate(students)))

for idx, name in enumerate(students):
    print(f"{idx}번 학생: {name}")
for idx, name in enumerate(students, start=1):
    print(f"{idx}번 학생: {name}")
print(tuple(zip(students, scores)))
for name, score in zip(students, scores):
    print(f"{name}의 점수: {score}점")
print(list(enumerate(zip(students, scores), start=1)))

# TODO: enumerate와 zip을 조합하거나 활용하여 80점 이상 학생 정보 출력
# TODO: 예상 출력 "1번 짱구 학생 (점수: 85점)"
for idx, (name, score) in enumerate(zip(students, scores), start=1):
    print(f"{idx}번 {name} 학생 (점수: {score}점)")
````

````python
# 1. Array Creation
arr_1d = np.array([1, 2, 3, 4, 5, 6])
zeros_2d = np.zeros((2, 3))       # 2x3 matrix filled with 0.0
seq_arr = np.arange(0, 10, 2)     # Array from 0 to 8 with step 2
lin_arr = np.linspace(0, 1, 5)    # 5 evenly spaced values between 0 and 1

print("1D Array:", arr_1d)
print("Sequence Array:", seq_arr)
print("Linspace Array:", lin_arr)

# 2. Reshaping and Transposition
matrix_2d = arr_1d.reshape(2, 3)  # Convert (6,) to (2, 3)
transposed = matrix_2d.T          # Transpose matrix to (3, 2)

print("\nReshaped 2x3 Matrix:\n", matrix_2d)
print("Transposed 3x2 Matrix:\n", transposed)

# 3. Multi-dimensional Slicing [rows, cols]
# Extract all rows, column index 1 to 2
slice_col = matrix_2d[:, 1:3]
# Extract specific element at row 0, col 2
element = matrix_2d[0, 2]

print("\nSliced Columns (col 1 to 2):\n", slice_col)
print("Element at (0, 2):", element)

# 4. Broadcasting (Element-wise Operations)
# Add a scalar to all elements
broadcast_scalar = matrix_2d + 10
# Add a 1D array to each row of a 2D matrix
broadcast_row = matrix_2d + np.array([100, 200, 300])

print("\nBroadcast with Scalar (+10):\n", broadcast_row)

# 5. Axis-based Aggregations
total_sum = np.sum(matrix_2d)              # Sum of all elements
col_sum = np.sum(matrix_2d, axis=0)        # Sum along vertical axis (column-wise)
row_mean = np.mean(matrix_2d, axis=1)      # Mean along horizontal axis (row-wise)

print("\nTotal Sum:", total_sum)
print("Sum along Axis 0 (Columns):", col_sum)
print("Mean along Axis 1 (Rows):", row_mean)
````

````python
# 1. Data Type Specification and Type Casting (.astype)
# Create float64 array by default
float_arr = np.array([0.0, 128.6, 255.4])
print("Original dtype:", float_arr.dtype)

# Convert to uint8 (commonly used for 8-bit image processing)
uint8_arr = float_arr.astype(np.uint8)
print("Converted array (uint8):", uint8_arr)
print("Converted dtype:", uint8_arr.dtype)

# 2. Element-wise Arithmetic Operations
arr_a = np.array([10, 20, 30])
arr_b = np.array([1, 2, 3])

print("\n=== Element-wise Operations ===")
print("Addition:", arr_a + arr_b)          # [11, 22, 33]
print("Multiplication:", arr_a * arr_b)    # [10, 40, 90]
print("Power:", arr_a ** 2)                # [100, 400, 900]

# 3. Matrix Multiplication (Linear Algebra)
mat_a = np.array([[1, 2], [3, 4]])
mat_b = np.array([[5, 6], [7, 8]])

# Matrix product using @ operator
matmul_result = mat_a @ mat_b
print("\n=== Matrix Multiplication (@) ===\n", matmul_result)

# 4. Comparison and Boolean Masking
data = np.array([15, 3, 27, 8, 42])
mask = data >= 15

print("\n=== Boolean Masking ===")
print("Mask (data >= 15):", mask)
print("Filtered Elements:", data[mask])     # Extract only True elements

# Conditional replacement using np.where(condition, if_true, if_false)
replaced = np.where(data >= 15, 999, 0)
print("Replaced with np.where:", replaced)

# 5. Statistical and Index Aggregation
print("\n=== Statistics and Index Search ===")
print("Index of Maximum Value (argmax):", np.argmax(data))
print("Standard Deviation:", np.std(data))
````
